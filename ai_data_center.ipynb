```python
import os
import pandas as pd

# Install tabulate jika belum ada untuk tampilan tabel rapi di console
!pip install tabulate -q
from tabulate import tabulate

# Pengaturan Pandas agar seluruh kolom dan teks terlihat rapi tanpa terpotong
pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', None)
pd.set_option('display.width', 1000)

print("=" * 80)
print(" SISTEM PENGOLAHAN DATASET AI DATA CENTER (RINGKASAN TABEL RAPI)")
print("=" * 80 + "\n")

# 1. BACA SELURUH DATASET CSV
files = {
    'data_centers': 'data_centers.csv',
    'chip_types': 'chip_types.csv',
    'chip_quantities': 'data_center_chip_quantities.csv',
    'chillers': 'data_center_chillers.csv',
    'cooling_towers': 'data_center_cooling_towers.csv',
    'timelines': 'data_center_timelines.csv'
}

dfs = {}
for key, path in files.items():
    if os.path.exists(path):
        dfs[key] = pd.read_csv(path)
    else:
        print(f"⚠️ Warning: File {path} tidak ditemukan!")

# -----------------------------------------------------------------------------
# TABEL 1: RINGKASAN DATA CENTER (20 Terbesar Berdasarkan Kapasitas Compute H100)
# -----------------------------------------------------------------------------
if 'data_centers' in dfs:
    print("\n📌 [TABEL 1] RINGKASAN AI DATA CENTER UTAMA")
    df_dc = dfs['data_centers'].copy()

    # Pilih kolom-kolom vital
    cols_dc = [
        'Name', 'Owner', 'Country', 'Current H100 equivalents',
        'Current power (MW)', 'Current total capital cost (2025 USD billions)', 'Current chip types'
    ]
    df_dc_clean = df_dc[cols_dc].copy()

    # Format angka agar mudah dibaca
    df_dc_clean['Current H100 equivalents'] = df_dc_clean['Current H100 equivalents'].apply(lambda x: f"{x:,.0f}" if pd.notnull(x) else "-")
    df_dc_clean['Current power (MW)'] = df_dc_clean['Current power (MW)'].apply(lambda x: f"{x:,.1f}" if pd.notnull(x) else "-")
    df_dc_clean['Current total capital cost (2025 USD billions)'] = df_dc_clean['Current total capital cost (2025 USD billions)'].apply(lambda x: f"${x:,.2f} B" if pd.notnull(x) else "-")

    # Rename header ke Bahasa Indonesia
    df_dc_clean.columns = ['Nama Data Center', 'Pemilik (Owner)', 'Negara', 'Kapasitas (H100 Equiv.)', 'Daya (MW)', 'Estimasi Biaya ($)', 'Tipe Chip']

    # Tampilkan 20 teratas
    print(tabulate(df_dc_clean.head(20), headers='keys', tablefmt='github', showindex=False))

# -----------------------------------------------------------------------------
# TABEL 2: SPESIFIKASI SPESIFIK & PERFORMA CHIP/GPU AI
# -----------------------------------------------------------------------------
if 'chip_types' in dfs:
    print("\n\n📌 [TABEL 2] DAFAR TIPE CHIP / GPU AI & SPESIFIKASI TEKNIS")
    df_ct = dfs['chip_types'].copy()

    cols_ct = ['Name', 'Designer', 'TDP (W) (from ML Hardware (linked))', 'Memory', 'Memory bandwidth', 'H100e', 'Cost per chip (approx.)', 'Release date']
    df_ct_clean = df_ct[cols_ct].copy()

    # Formatting
    df_ct_clean['Memory'] = df_ct_clean['Memory'].apply(lambda x: f"{x/1e9:.1f} GB" if pd.notnull(x) and x > 0 else "-")
    df_ct_clean['Memory bandwidth'] = df_ct_clean['Memory bandwidth'].apply(lambda x: f"{x/1e12:.2f} TB/s" if pd.notnull(x) and x > 0 else "-")
    df_ct_clean['Cost per chip (approx.)'] = df_ct_clean['Cost per chip (approx.)'].apply(lambda x: f"${x:,.0f}" if pd.notnull(x) else "-")
    df_ct_clean['H100e'] = df_ct_clean['H100e'].apply(lambda x: f"{x:.2f}" if pd.notnull(x) else "-")

    df_ct_clean.columns = ['Nama Chip', 'Perancang', 'TDP (Watt)', 'Memori', 'Bandwidth Memori', 'Faktor Skala H100', 'Estimasi Harga/Chip', 'Tanggal Rilis']

    print(tabulate(df_ct_clean.dropna(subset=['Nama Chip']).head(20), headers='keys', tablefmt='github', showindex=False))

# -----------------------------------------------------------------------------
# TABEL 3: REKAPITULASI TOTAL ALOKASI CHIP BERDASARKAN PEMILIK / PENGGUNA
# -----------------------------------------------------------------------------
if 'chip_quantities' in dfs:
    print("\n\n📌 [TABEL 3] REKAPITULASI PENGGUNAAN CHIP PADA BERBAGAI PROYEK DATA CENTER")
    df_cq = dfs['chip_quantities'].copy()

    # Agregasi jumlah unit chip per Data Center dan Chip Type
    rekap_chip = df_cq.groupby(['Data center', 'Owner', 'Chip type'])['Number of Units'].sum().reset_index()
    rekap_chip = rekap_chip.sort_values(by='Number of Units', ascending=False)

    rekap_chip['Number of Units'] = rekap_chip['Number of Units'].apply(lambda x: f"{x:,.0f}")
    rekap_chip.columns = ['Nama Data Center', 'Pemilik', 'Tipe Chip', 'Total Jumlah Unit']

    print(tabulate(rekap_chip.head(20), headers='keys', tablefmt='github', showindex=False))

# -----------------------------------------------------------------------------
# TABEL 4: SPESIFIKASI INFRASTRUKTUR PENDINGIN (CHILLERS & COOLING TOWERS)
# -----------------------------------------------------------------------------
if 'chillers' in dfs:
    print("\n\n📌 [TABEL 4] RINGKASAN SISTEM PENDINGIN (CHILLERS)")
    df_ch = dfs['chillers'].copy()

    cols_ch = ['Model', 'Manufacturer', 'Type', 'Cooling capacity (kW)', 'Number of fans', 'Area (m^2)']
    df_ch_clean = df_ch[cols_ch].copy()
    df_ch_clean.columns = ['Model Chiller', 'Produsen', 'Tipe', 'Kapasitas Pendinginan (kW)', 'Jumlah Kipas', 'Luas Area (m²)']

    print(tabulate(df_ch_clean.head(10), headers='keys', tablefmt='github', showindex=False))
```

# 2. KLASIFIKASI BERDASARKAN GEOGRAFI

```python
import os
import pandas as pd

# Pengaturan visualisasi tabel di Google Colab agar rapi dan lengkap
pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', None)
pd.set_option('display.width', 1000)

print("==========================================================================================")
print("  PEMETAAN RISIKO KRISIS AIR BERSIH OPERASIONAL AI DATA CENTER GLOBAL (DATASET EPOCH AI)")
print("==========================================================================================\n")

# 1. Pemuatan Dataset
if os.path.exists('data_centers.csv') and os.path.exists('data_center_timelines.csv'):
    df_dc = pd.read_csv('data_centers.csv')
    df_tl = pd.read_csv('data_center_timelines.csv')

    # Ambil angka konsumsi air terbaru (Water use MGD) per Data Center dari file timelines
    df_tl_water = df_tl.dropna(subset=['Water use (MGD)']).sort_values(by='Date', ascending=False)
    df_water_latest = df_tl_water.groupby('Data center')['Water use (MGD)'].first().reset_index()

    # Gabungkan (Merge) data utama dengan konsumsi air
    df_merged = pd.merge(df_dc, df_water_latest, left_on='Name', right_on='Data center', how='left')

    # 2. Fungsi Klasifikasi Wilayah Operasional (Gunung, Padat Penduduk, Bawah Laut/Pesisir)
    def klasifikasi_wilayah(row):
        name = str(row['Name']).lower()
        addr = str(row['Address']).lower()
        country = str(row['Country']).lower()
        text = f"{name} {addr} {country}"

        # Indikator Bawah Laut / Pesisir / Offshore / Tepi Laut
        if any(kw in text for kw in ['ocean', 'sea', 'subsea', 'offshore', 'coastal', 'lake', 'mariner', 'bay', 'water']):
            return 'Bawah Laut / Pesisir / Tepi Air'
        # Indikator Gunung / Dataran Tinggi / Pegunungan
        elif any(kw in text for kw in ['mountain', 'mount', 'hill', 'cheyenne', 'highland', 'ridge', 'alps', 'nm', 'utah']):
            return 'Pegunungan / Dataran Tinggi'
        # Indikator Padat Penduduk / Perkotaan / Metropolitan
        elif any(kw in text for kw in ['atlanta', 'phoenix', 'columbus', 'chicago', 'johor bahru', 'omaha', 'richmond', 'los lunas', 'singapore']):
            return 'Padat Penduduk / Perkotaan'
        else:
            return 'Wilayah Daratan / Suburb'

    df_merged['Tipe Wilayah Geografis'] = df_merged.apply(klasifikasi_wilayah, axis=1)

    # 3. Kuantifikasi Penggunaan Air & Estimasi Risiko
    # 1 MGD (Million Gallons per Day) = 1,000,000 Galon/hari ≈ 3,785,412 Liter/hari
    def hitung_air_galon(mgd):
        if pd.notnull(mgd) and mgd > 0:
            return f"{mgd * 1_000_000:,.0f} Galon/hari"
        return "Sistem Pendingin Tanpa Air / Closed-Loop"

    def hitung_air_liter(mgd):
        if pd.notnull(mgd) and mgd > 0:
            return f"{mgd * 3_785_412:,.0f} Liter/hari"
        return "0 Liter/hari (Air Ditaur Ulang)"

    def estimasi_tingkat_risiko(row):
        water = row['Water use (MGD)']
        power = row['Current power (MW)']
        wilayah = row['Tipe Wilayah Geografis']

        if wilayah == 'Padat Penduduk / Perkotaan':
            if (pd.notnull(water) and water > 0.5) or (pd.notnull(power) and power > 200):
                return 'TINGGI (Kompetisi Air dengan Warga)'
            return 'SEDANG (Beban Infrastruktur Kota)'
        elif wilayah == 'Pegunungan / Dataran Tinggi':
            if pd.notnull(water) and water > 0:
                return 'TINGGI (Kelangkaan Sumber Air Alami)'
            return 'SEDANG (Pendataran & Ekosistem Terbatas)'
        elif wilayah == 'Bawah Laut / Pesisir / Tepi Air':
            return 'RENDAH - SEDANG (Termal & Korosif Garam)'
        else:
            if pd.notnull(water) and water > 1.0:
                return 'TINGGI (Pengambilan Air Tanah Masif)'
            return 'SEDANG'

    df_merged['Estimasi Penggunaan Air (Galon)'] = df_merged['Water use (MGD)'].apply(hitung_air_galon)
    df_merged['Estimasi Penggunaan Air (Liter)'] = df_merged['Water use (MGD)'].apply(hitung_air_liter)
    df_merged['Tingkat Risiko Krisis Air'] = df_merged.apply(estimasi_tingkat_risiko, axis=1)

    # 4. Filter & Menampilkan Tabel Utama Berdasarkan Kriteria Wilayah Spesifik
    kolom_tampil = [
        'Name', 'Owner', 'Country', 'Tipe Wilayah Geografis',
        'Current power (MW)', 'Estimasi Penggunaan Air (Liter)',
        'Estimasi Penggunaan Air (Galon)', 'Tingkat Risiko Krisis Air'
    ]

    df_final = df_merged[kolom_tampil].copy()

    # Format Daya (MW)
    df_final['Current power (MW)'] = df_final['Current power (MW)'].apply(lambda x: f"{x:,.1f} MW" if pd.notnull(x) else "Tidak Terdata")

    # Rename Kolom ke Bahasa Indonesia
    df_final.columns = [
        'Nama AI Data Center', 'Pemilik / Pengelola', 'Negara',
        'Karakteristik Wilayah', 'Kapasitas Daya (MW)',
        'Konsumsi Air (Liter/Hari)', 'Konsumsi Air (Galon/Hari)', 'Tingkat Risiko Krisis Air'
    ]

    print("=== TABEL 1: PEMETAAN RISIKO KRISIS AIR BERSIH PADA KATEGORI WILAYAH KHUSUS ===")
    df_spesifik = df_final[df_final['Karakteristik Wilayah'].isin([
        'Pegunungan / Dataran Tinggi',
        'Padat Penduduk / Perkotaan',
        'Bawah Laut / Pesisir / Tepi Air'
    ])].sort_values(by='Karakteristik Wilayah').reset_index(drop=True)

    display(df_spesifik)

    print("\n" + "="*100 + "\n")

    print("=== TABEL 2: REKAPITULASI RINGKASAN RISIKO AIR BERDASARKAN TIPE WILAYAH ===")
    rekap = df_merged.groupby('Tipe Wilayah Geografis').agg(
        Jumlah_Data_Center=('Name', 'count'),
        Rata_Daya_MW=('Current power (MW)', 'mean'),
        Total_Air_MGD=('Water use (MGD)', 'sum')
    ).reset_index()

    rekap['Total Konsumsi Air (Liter/Hari)'] = rekap['Total_Air_MGD'].apply(lambda x: f"{x * 3_785_412:,.0f} Liter/hari")
    rekap['Rata-rata Daya (MW)'] = rekap['Rata_Daya_MW'].apply(lambda x: f"{x:,.1f} MW" if pd.notnull(x) else "-")

    rekap_tampil = rekap[['Tipe Wilayah Geografis', 'Jumlah_Data_Center', 'Rata-rata Daya (MW)', 'Total Konsumsi Air (Liter/Hari)']]
    rekap_tampil.columns = ['Tipe Wilayah Operasional', 'Jumlah Fasilitas', 'Rata-Rata Kebutuhan Daya', 'Total Konsumsi Air Bersih']

    display(rekap_tampil)

else:
    print("File dataset 'data_centers.csv' atau 'data_center_timelines.csv' tidak ditemukan. Harap pastikan file sudah diunggah.")
```

# 3. KLASIFIKASI BERDASARKAN GEOGRAFI (2)

```python
import os
import pandas as pd
import numpy as np

# Pengaturan Tampilan Pandas agar Tabel Rapi & Tidak Terpotong
pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', None)
pd.set_option('display.width', 1000)

print("=========================================================================================")
print("  PROGRAM PEMETAAN RISIKO KRISIS AIR BERSIH OPERASIONAL AI DATA CENTER GLOBAL (EPOCH AI)")
print("=========================================================================================\n")

# 1. PEMUATAN DATASET
if not (os.path.exists('data_centers.csv') and os.path.exists('data_center_timelines.csv')):
    print("❌ Error: File 'data_centers.csv' atau 'data_center_timelines.csv' tidak ditemukan!")
else:
    df_dc = pd.read_csv('data_centers.csv')
    df_tl = pd.read_csv('data_center_timelines.csv')

    # Mengambil data konsumsi air terbaru dari timeline jika tersedia (dalam MGD - Million Gallons per Day)
    water_latest = df_tl.groupby('Data center')['Water use (MGD)'].last().to_dict()

    # 2. KLASIFIKASI TIPE WILAYAH OPERASIONAL & PERHITUNGAN KONSUMSI AIR
    def klasifikasi_wilayah(row):
        text = (str(row['Name']) + " " + str(row['Address']) + " " + str(row['Country'])).lower()

        # Indikator Bawah Laut / Pesisir / Subsea
        if any(k in text for k in ['underwater', 'subsea', 'ocean', 'offshore', 'aquatic', 'marine', 'coastal']):
            return 'Bawah Laut / Pesisir'
        # Indikator Pegunungan & Dataran Tinggi (Geografi Tinggi / Lembah / Daerah Berbukit)
        elif any(k in text for k in ['mountain', 'hill', 'peak', 'highland', 'ridge', 'sierra', 'utah', 'colorado', 'cheyenne', 'sandia', 'prineville', 'quincy']):
            return 'Pegunungan & Dataran Tinggi'
        # Default: Wilayah Perkotaan & Industri Padat Penduduk
        else:
            return 'Padat Penduduk / Perkotaan'

    def hitung_air_mgd(row):
        name = row['Name']
        # Gunakan data historis terverifikasi dari timeline jika ada
        w = water_latest.get(name, np.nan)
        if pd.notnull(w) and w > 0:
            return w
        # Jika tidak ada, gunakan estimasi konsumsi sistem pendingin air standar (rata-rata 0.0035 MGD per MW daya)
        p = row['Current power (MW)']
        if pd.notnull(p) and p > 0:
            return p * 0.0035
        return 0.0

    df_dc['Tipe Wilayah Operasional'] = df_dc.apply(klasifikasi_wilayah, axis=1)
    df_dc['Air_MGD'] = df_dc.apply(hitung_air_mgd, axis=1)

    # Konversi MGD (Juta Galon/Hari) ke Liter per Hari untuk kemudahan pemahaman masyarakat umum (1 Galon US = 3.78541 Liter)
    df_dc['Air_Liter_Hari'] = df_dc['Air_MGD'] * 1_000_000 * 3.78541

    # Skor & Klasifikasi Risiko Air
    def kalkulasi_risiko(row):
        tw = row['Tipe Wilayah Operasional']
        daya = row['Current power (MW)'] if pd.notnull(row['Current power (MW)']) else 0

        if tw == 'Padat Penduduk / Perkotaan':
            if daya >= 500: return 'Sangat Tinggi (Kritis)'
            elif daya >= 200: return 'Tinggi'
            else: return 'Sedang'
        elif tw == 'Pegunungan & Dataran Tinggi':
            if daya >= 300: return 'Tinggi (Ancaman Sumber Air Hulu)'
            else: return 'Sedang'
        else:
            return 'Rendah (Pendinginan Laut Natif)'

    df_dc['Tingkat Risiko Air'] = df_dc.apply(kalkulasi_risiko, axis=1)

    # -----------------------------------------------------------------------------------------
    # TABEL 1: REKAPITULASI RINGKASAN RISIKO KRISIS AIR BERDASARKAN TIPE WILAYAH OPERASIONAL
    # -----------------------------------------------------------------------------------------
    print("=== TABEL 1: REKAPITULASI AGREGAT KRISIS AIR BERDASARKAN TIPE WILAYAH OPERASIONAL ===")

    rekap_wilayah = df_dc.groupby('Tipe Wilayah Operasional').agg(
        Jumlah_Fasilitas=('Name', 'count'),
        Total_Daya_MW=('Current power (MW)', 'sum'),
        Total_Air_MGD=('Air_MGD', 'sum'),
        Total_Air_Liter_Hari=('Air_Liter_Hari', 'sum')
    ).reset_index()

    # Format Tampilan Angka agar Mudah Dibaca
    rekap_wilayah['Total Daya (MW)'] = rekap_wilayah['Total_Daya_MW'].apply(lambda x: f"{x:,.1f} MW")
    rekap_wilayah['Total Konsumsi Air (MGD)'] = rekap_wilayah['Total_Air_MGD'].apply(lambda x: f"{x:,.2f} MGD")
    rekap_wilayah['Total Konsumsi Air Bersih (Liter/Hari)'] = rekap_wilayah['Total_Air_Liter_Hari'].apply(lambda x: f"{x:,.0f} L/hari")

    tabel_1_final = rekap_wilayah[[
        'Tipe Wilayah Operasional', 'Jumlah_Fasilitas',
        'Total Daya (MW)', 'Total Konsumsi Air (MGD)', 'Total Konsumsi Air Bersih (Liter/Hari)'
    ]]
    tabel_1_final.columns = [
        'Tipe Wilayah Operasional', 'Jumlah Fasilitas',
        'Total Kebutuhan Daya', 'Total Konsumsi Air (MGD)', 'Total Konsumsi Air Bersih'
    ]

    display(tabel_1_final)

    # -----------------------------------------------------------------------------------------
    # TABEL 2: REKAPITULASI RISIKO AIR TERPARAH (TOP KRITIS) PER TIPE WILAYAH OPERASIONAL
    # -----------------------------------------------------------------------------------------
    print("\n=========================================================================================")
    print("=== TABEL 2: REKAPITULASI FASILITAS DENGAN ANCAMAN RISIKO AIR TERPARAH PER TIPE WILAYAH ===")
    print("=========================================================================================")

    # Ambil Fasilitas Berisiko Terparah dari Setiap Tipe Wilayah
    top_kritis = df_dc.sort_values(by=['Air_MGD', 'Current power (MW)'], ascending=False) \
                      .groupby('Tipe Wilayah Operasional').head(5).reset_index(drop=True)

    # Pembersihan Nama Pemilik
    top_kritis['Pemilik Clean'] = top_kritis['Owner'].astype(str).str.replace(r'#\w+', '', regex=True).str.strip()

    # Formating Kolom
    top_kritis['Kebutuhan Daya'] = top_kritis['Current power (MW)'].apply(lambda x: f"{x:,.1f} MW" if pd.notnull(x) else "-")
    top_kritis['Konsumsi Air (MGD)'] = top_kritis['Air_MGD'].apply(lambda x: f"{x:,.2f} MGD")
    top_kritis['Konsumsi Air Bersih'] = top_kritis['Air_Liter_Hari'].apply(lambda x: f"{x:,.0f} L/hari")

    tabel_2_final = top_kritis[[
        'Tipe Wilayah Operasional', 'Name', 'Pemilik Clean', 'Country',
        'Kebutuhan Daya', 'Konsumsi Air (MGD)', 'Konsumsi Air Bersih', 'Tingkat Risiko Air'
    ]]

    tabel_2_final.columns = [
        'Tipe Wilayah Operasional', 'Nama Data Center', 'Pemilik / Pengelola', 'Negara',
        'Kebutuhan Daya', 'Konsumsi Air (MGD)', 'Total Konsumsi Air Bersih (Liter/Hari)', 'Tingkat Risiko Air'
    ]

    display(tabel_2_final)

    # -----------------------------------------------------------------------------------------
    # TABEL 3: AGREGASI RISIKO AIR BERDASARKAN NEGARA LOKASI OPERASIONAL
    # -----------------------------------------------------------------------------------------
    print("\n=========================================================================================")
    print("=== TABEL 3: DISTRIBUSI BEBAN KONSUMSI AIR BERSIH AI DATA CENTER BERDASARKAN NEGARA ===")
    print("=========================================================================================")

    rekap_negara = df_dc.groupby('Country').agg(
        Jumlah_Fasilitas=('Name', 'count'),
        Total_Daya=('Current power (MW)', 'sum'),
        Total_Air_Liter=('Air_Liter_Hari', 'sum')
    ).sort_values(by='Total_Air_Liter', ascending=False).reset_index()

    rekap_negara['Total Kebutuhan Daya'] = rekap_negara['Total_Daya'].apply(lambda x: f"{x:,.1f} MW")
    rekap_negara['Total Konsumsi Air Bersih'] = rekap_negara['Total_Air_Liter'].apply(lambda x: f"{x:,.0f} L/hari")

    tabel_3_final = rekap_negara[['Country', 'Jumlah_Fasilitas', 'Total Kebutuhan Daya', 'Total Konsumsi Air Bersih']]
    tabel_3_final.columns = ['Negara Lokasi Operasional', 'Jumlah Fasilitas', 'Total Kebutuhan Daya', 'Total Konsumsi Air Bersih (Liter/Hari)']

    display(tabel_3_final.head(10))
```
